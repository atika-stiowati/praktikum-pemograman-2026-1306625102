```python
# Permainan Tebak Nama Buah
# Oleh : Atika Stiowati
# NIM  : 1306625102

import random

databuah = ['anggur', 'apel', 'aprikot', 'asam', 'atap', 'arbei', 'abiu', 'alpukat',
    'anjeng', 'bacang', 'baguk', 'belimbing', 'bengkuang', 'benda', 'binjai', 'bisbul',
    'blueberry', 'burahol', 'blewah', 'bit', 'cempedak', 'ceplukan', 'cermai', 'ceri',
    'cokelat', 'delima', 'duku', 'durian', 'duwet', 'enau', 'enamenam', 'frambos',
    'hamoi', 'hamlam', 'holdi', 'jamblang', 'jambuair', 'jambubatu', 'jambubol',
    'jambumawar', 'jambumede', 'jengkol', 'jeruk', 'jerukbali', 'jerukkeprok',
    'jerukkingkit', 'jeruknipis', 'jerukpurut', 'kapulasan', 'kawista', 'kecapi',
    'kedondong', 'kelapa', 'kelengkeng', 'kelubi', 'ketela', 'kemang', 'kepel',
    'kersen', 'kesemek', 'kokosan', 'kiwi', 'koldi', 'kurma', 'kelawi', 'kenitu',
    'lai', 'langsat', 'lemon', 'leci', 'lobak', 'maja', 'malaka', 'mangga', 'manggis',
    'markisa', 'melon', 'mengkudu', 'menteng', 'matoa', 'mentimun', 'namnam', 'nanas',
    'nangka', 'naga', 'nona', 'pepaya', 'persik', 'pir', 'pisang', 'pete', 'rambai',
    'rambusa', 'rambutan', 'rukem', 'sawo', 'semangka', 'sirsak', 'siwalan', 'srikaya',
    'stroberi', 'sukun', 'terap', 'terong', 'tomat', 'timunsuri', 'ubi', 'waluh',
    'wuni', 'zaitun']


def tampil_tebakan(buah, benar):
    """Mengubah nama buah menjadi pola, huruf yang belum tertebak = '-'."""
    return ' '.join(h if h in benar else '-' for h in buah)


def main_satu_ronde():
    buah = random.choice(databuah)      # komputer memilih buah secara acak
    nyawa = len(buah)                   # nyawa = jumlah huruf
    benar = set()                       # huruf yang sudah benar
    salah = set()                       # huruf yang sudah salah

    print(f"\nKomputer sedang memikirkan nama buah terdiri dari {len(buah)} huruf.")
    print("Nama buah :", tampil_tebakan(buah, benar))

    while nyawa > 0 and set(buah) != benar:
        print(f"Anda memiliki {nyawa} kesempatan untuk menebak huruf buah tersebut.")
        tebak = input("Tebakan anda ? ").strip().lower()

        # validasi input
        if len(tebak) != 1 or not tebak.isalpha():
            print("Masukkan satu huruf saja.")
            continue
        if tebak in benar or tebak in salah:
            print(f"Huruf {tebak} sudah pernah ditebak.")
            continue

        if tebak in buah:
            benar.add(tebak)
            print(f"Benar. Huruf {tebak} ada dalam nama buah tersebut.")
        else:
            salah.add(tebak)
            nyawa -= 1
            print(f"Sayang sekali. Huruf {tebak} tidak ada dalam nama buah tersebut.")
        print("Nama buah :", tampil_tebakan(buah, benar))

    if set(buah) == benar:
        print("SELAMAT. Anda berhasil !!")
    else:
        print("Anda kalah !!")
        print("Nama buah tersebut :", ' '.join(buah))


def main():
    print("Permainan Tebak Nama Buah")
    print("Oleh : Atika Stiowati")
    print("NIM  : 1306625102")
    while True:
        main_satu_ronde()
        ulang = input("\nMau main lagi (y/t) ? ").strip().lower()
        if ulang != 'y':
            break
    print("# SELESAI #")


if __name__ == "__main__":
    main()

```
